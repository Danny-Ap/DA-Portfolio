```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go
```

```python
# Get top countries list of most players in Serie A
pl_num = pd.read_csv("PL_Num_Of_Players.csv")
pl_num.set_index('Unnamed: 0', inplace=True)
pl_num.rename_axis("Country", inplace=True)
pl_num_top = pl_num.sum(axis=1).sort_values(ascending=False)
pl_num_top_list = pl_num_top.index.tolist()


# Get top countries list of most minutes played by country in Serie A
pl_min = pd.read_csv("PL_Min_Played.csv")
pl_min.set_index('Unnamed: 0', inplace=True)
pl_min.rename_axis("Country", inplace=True)
pl_min_top = pl_min.sum(axis=1).sort_values(ascending=False)
pl_min_top_list = pl_min_top.index.tolist()
```

```python
# Get total number of players per year percentage
pl_num_total = pl_num.sum(axis=0)
pl_num_percentage = pl_num.copy()
for year in pl_num.columns:
    pl_num_percentage[year] = (pl_num[year] / pl_num_total[year] * 100).round(2)

pl_num_percentage
```

```python
# Get total minutes played per year percentage
pl_min_total = pl_min.sum(axis=0)
pl_min_percentage = pl_min.copy()
for year in pl_min.columns:
    pl_min_percentage[year] = (pl_min[year] / pl_min_total[year] * 100).round(2)

pl_min_percentage
```

```python
# Select the top 10 countries (skip England, 1st row)
pl_num_top_10 = pl_num.loc[pl_num_top_list[1:11]]

# Plot each country's data
plt.figure(figsize=(12, 8))
for index, row in pl_num_top_10.iterrows():
    plt.plot(row.index, row.values, label=index)
    plt.annotate(index, xy=(row.index[-1], row.values[-1]), xytext=(5, 0), textcoords='offset points')

# Add labels and title
plt.xlabel('Year')
plt.xticks(rotation=90)
plt.ylabel('Number of Players')
plt.title('Number of Players by Country Over Years')
plt.grid(axis='both',linestyle='--', alpha=0.7)
plt.yticks(range(0, int(pl_num_top_10.max().max())+1, 5))

# Show the plot
plt.show()
```

```python
# Select the top 10 countries (skip Italy, 1st row)
pl_num_percentage_top_10 = pl_num_percentage.loc[pl_num_top_list[1:11]]

# Plot each country's data
plt.figure(figsize=(12, 8))
for index, row in pl_num_percentage_top_10.iterrows():
    plt.plot(row.index, row.values, label=index)
    plt.annotate(index, xy=(row.index[-1], row.values[-1]), xytext=(5, 0), textcoords='offset points')

# Add labels and title
plt.xlabel('Year')
plt.xticks(rotation=90)
plt.ylabel('Percentage')
plt.title('Percentage of Foreign Players by Country Over Years')
plt.grid(axis='both',linestyle='--', alpha=0.7)
plt.yticks(range(0, int(pl_num_percentage_top_10.max().max())+1, 1))

# Show the plot
plt.show()
```

```python
df1 = pd.read_csv('PL_Num_Of_Players.csv')
df1.set_index('Unnamed: 0', inplace=True)
df1.loc['Total Players'] = df1.sum()
pl_total_num_of_players = df1.loc[['Total Players']]

df2 = pd.read_csv('PL_Num_Of_Players.csv')
df2.set_index('Unnamed: 0', inplace=True)
df2.loc['Foreign Players'] = df2.iloc[1:].sum()
pl_foreign_players_number = df2.loc[['Foreign Players']]

pl_foreign_percentage = pl_total_num_of_players.copy()
pl_foreign_percentage.loc['Foreign Players'] = pl_foreign_players_number.iloc[0]
percentage_row = (pl_foreign_percentage.loc['Foreign Players'] / pl_foreign_percentage.loc['Total Players'] * 100).round(2)
pl_foreign_percentage.loc['Percentage'] = percentage_row
pl_foreign_percentage
```

```python
# Cleaning df for worldmap scatter plot

pl_foreign_countries = pd.read_csv("PL_Num_Of_Players.csv")
pl_foreign_countries.set_index('Unnamed: 0', inplace=True)
pl_foreign_countries.rename_axis("Country", inplace=True)

pl_foreign_countries = pl_foreign_countries.iloc[1:]  # Remove England row
pl_foreign_countries.index = pl_foreign_countries.index.str.extract(r'([A-Z][a-zA-Z\s]*)$')[0]  # Extract country names

pl_foreign_countries_long = pl_foreign_countries.reset_index(names='Country').melt(
    id_vars='Country', var_name='Year', value_name='Players'
)

pl_foreign_countries_long
```

```python
# Create the base figure
fig = go.Figure()

# Add the initial choropleth map for the first year
initial_year = pl_foreign_countries_long['Year'].min()
filtered_data = pl_foreign_countries_long[pl_foreign_countries_long['Year'] == initial_year]

fig.add_trace(
    go.Choropleth(
        locations=filtered_data['Country'],
        locationmode="country names",
        z=filtered_data['Players'],
        text=filtered_data['Country'],
        colorscale=["white","red"],
        colorbar_title="Players",
        showscale=False,
    )
)

# Add frames for each year
years = pl_foreign_countries_long['Year'].unique()
frames = []
for year in years:
    year_data = pl_foreign_countries_long[pl_foreign_countries_long['Year'] == year]
    
    # Filter for UK countries
    uk_countries = year_data[year_data['Country'].isin(['England', 'Scotland', 'Wales', 'Northern Ireland'])]
    uk_text = "<b>UK countries:</b><br>"
    for _, row in uk_countries.iterrows():
        uk_text += f"{row['Country']}: <span style='float:right'>{int(row['Players'])}</span><br>"
    
    annotations = [
        # Right box - foreign players info
        dict(
            x=1.05,
            y=1,
            xref="paper",
            yref="paper",
            text=(
                f"<b>Year: {year}</b><br>"
                f"Total Players: <span style='float:right'>{int(pl_foreign_percentage.loc['Total Players', year])}</span><br>"
                f"Foreign Players: <span style='float:right'>{int(pl_foreign_percentage.loc['Foreign Players', year])}</span><br>"
                f"Percentage: <span style='float:right'>{pl_foreign_percentage.loc['Percentage', year]:.2f}%</span>"
            ),
            showarrow=False,
            align="left",
            font=dict(family="Arial", size=14, color="black"),
            bgcolor="lightgrey",
            bordercolor="black",
            borderwidth=1,
        ),
        # Left box - UK countries info
        dict(
            x=-0.05,
            y=1,
            xref="paper",
            yref="paper",
            text=uk_text,
            showarrow=False,
            align="left",
            font=dict(family="Arial", size=14, color="black"),
            bgcolor="lightgrey",
            bordercolor="black",
            borderwidth=1,
        )
    ]
    frames.append(
        go.Frame(
            data=[
                go.Choropleth(
                    locations=year_data['Country'],
                    locationmode="country names",
                    z=year_data['Players'],
                    text=year_data['Country'],
                    colorscale=["white","red"],
                    showscale=False,
                )
            ],
            name=str(year),
            layout=go.Layout(annotations=annotations),
        )
    )

fig.frames = frames

# Initial annotations for the first display
initial_uk_countries = filtered_data[filtered_data['Country'].isin(['Scotland', 'Wales', 'Northern Ireland'])]
initial_uk_text = "<b>UK countries:</b><br>"
for _, row in initial_uk_countries.iterrows():
    initial_uk_text += f"{row['Country']}: <span style='float:right'>{int(row['Players'])}</span><br>"

# Update layout for better visuals
fig.update_layout(
    geo=dict(
        showframe=False,
        showcoastlines=True,
        showcountries=True,
        coastlinecolor="black",
        countrycolor="black",
        oceancolor="white",
        bgcolor="white",
        projection_type="natural earth",
        resolution=110,
        fitbounds='locations',
        lakecolor="white",
    ),
    title=dict(
        text="<b>Number of foreign players in the Premier League by year</b>",
        x=0.5,
        font=dict(family="Arial", size=18),
    ),
    autosize=True,
    margin=dict(l=150, r=150, t=80, b=80),  # Increased left margin to accommodate the new box
    paper_bgcolor="white",
    font=dict(family="Arial"),
    annotations=[
        # Initial right box
        dict(
            x=1.05,
            y=1,
            xref="paper",
            yref="paper",
            text=(
                f"<b>Year: {initial_year}</b><br>"
                f"Total Players: <span style='float:right'>{int(pl_foreign_percentage.loc['Total Players', initial_year])}</span><br>"
                f"Foreign Players: <span style='float:right'>{int(pl_foreign_percentage.loc['Foreign Players', initial_year])}</span><br>"
                f"Percentage: <span style='float:right'>{pl_foreign_percentage.loc['Percentage', initial_year]:.2f}%</span>"
            ),
            showarrow=False,
            align="left",
            font=dict(family="Arial", size=14, color="black"),
            bgcolor="lightgrey",
            bordercolor="black",
            borderwidth=1,
        ),
        # Initial left box
        dict(
            x=-0.05,
            y=1,
            xref="paper",
            yref="paper",
            text=initial_uk_text,
            showarrow=False,
            align="left",
            font=dict(family="Arial", size=14, color="black"),
            bgcolor="lightgrey",
            bordercolor="black",
            borderwidth=1,
        )
    ],
    updatemenus=[
        dict(
            type="buttons",
            showactive=False,
            y=0,
            x=0.1,
            xanchor="right",
            yanchor="top",
            direction="left",
            buttons=[
                dict(label="▶", method="animate", args=[None, {"frame": {"duration": 500, "redraw": True}, "fromcurrent": True}]),
                dict(label="■", method="animate", args=[[None], {"frame": {"duration": 0, "redraw": False}, "mode": "immediate"}]),
            ],
        )
    ],
)

# Add slider
fig.update_layout(
    sliders=[
        dict(
            steps=[
                dict(
                    method="animate",
                    args=[[str(year)], {"frame": {"duration": 500, "redraw": True}, "mode": "immediate"}],
                    label=str(year),
                )
                for year in years
            ],
            transition={"duration": 300},
            x=0.15,
            xanchor="left",
            y=0,
            yanchor="top",
            pad={"t": 50, "b": 10},
            len=0.75,
        )
    ]
)

# Show the map
fig.show()
```

```python
# Create the base figure
fig = go.Figure()

# Find global maximum for consistent color scale
global_max = pl_foreign_countries_long['Players'].max()

# Add the initial choropleth map for the first year
initial_year = pl_foreign_countries_long['Year'].min()
filtered_data = pl_foreign_countries_long[pl_foreign_countries_long['Year'] == initial_year]

fig.add_trace(
    go.Choropleth(
        locations=filtered_data['Country'],
        locationmode="country names",
        z=filtered_data['Players'],
        text=filtered_data['Country'],
        colorscale="Reds",  # Changed from ["white","red"] to "Reds"
        colorbar_title="Players",
        zmin=0,
        zmax=global_max,  # Set global maximum
        showscale=True,    # Show the color scale
    )
)
fig.add_trace(
    go.Choropleth(
        locations=['United Kingdom'],  # Only England
        locationmode="country names",
        z=[0],  # Dummy value, since we're just highlighting
        showscale=False,
        marker=dict(
            line=dict(
                color='black',  # Border color
                width=2         # Border width
            )
        ),
        colorscale=[[0, 'rgba(255,255,0,0.3)'], [1, 'rgba(255,255,0,0.3)']],  # Yellow highlight with transparency
        name='UK Highlight'
    )
)

# Add frames for each year
years = pl_foreign_countries_long['Year'].unique()
frames = []
for year in years:
    year_data = pl_foreign_countries_long[pl_foreign_countries_long['Year'] == year]
    
    # Get UK countries data for this year
    uk_countries = year_data[year_data['Country'].isin(['Scotland', 'Wales', 'Northern Ireland'])]
    
    annotations = [
        dict(
            x=0.01,  # Moved to left side
            y=0.99,  # Moved to top
            xref="paper",
            yref="paper",
            text=(
                f"<b>Year: {year}</b><br>"
                f"Total Players: {int(pl_foreign_percentage.loc['Total Players', year])}<br>"
                f"Foreign Players: {int(pl_foreign_percentage.loc['Foreign Players', year])}<br>"
                f"Percentage: {pl_foreign_percentage.loc['Percentage', year]:.2f}%"
            ),
            showarrow=False,
            align="left",
            font=dict(size=14, color="black"),
            bgcolor="rgba(255, 255, 255, 0.8)",  # Semi-transparent background
            bordercolor="black",
            borderwidth=1,
        ),
        # Add UK countries box
        dict(
            x=0.01,  # Aligned with the Year box
            y=0.82,  # Positioned below the Year box
            xref="paper",
            yref="paper",
            text=(
                f"<b>UK countries:</b><br>"
                f"Scotland: {uk_countries[uk_countries['Country'] == 'Scotland']['Players'].sum() if not uk_countries[uk_countries['Country'] == 'Scotland'].empty else 0}<br>"
                f"Wales: {uk_countries[uk_countries['Country'] == 'Wales']['Players'].sum() if not uk_countries[uk_countries['Country'] == 'Wales'].empty else 0}<br>"
                f"Northern Ireland: {uk_countries[uk_countries['Country'] == 'Northern Ireland']['Players'].sum() if not uk_countries[uk_countries['Country'] == 'Northern Ireland'].empty else 0}"
            ),
            showarrow=False,
            align="left",
            font=dict(size=14, color="black"),
            bgcolor="rgba(255, 255, 255, 0.8)",  # Semi-transparent background
            bordercolor="black",
            borderwidth=1,
        )
    ]
    
    frames.append(
        go.Frame(
            data=[
                go.Choropleth(
                    locations=year_data['Country'],
                    locationmode="country names",
                    z=year_data['Players'],
                    text=year_data['Country'],
                    colorscale="Reds",  # Changed from ["white","red"] to "Reds"
                    zmin=0,
                    zmax=global_max,  # Set global maximum
                    showscale=True,
                ),
                go.Choropleth(
                    locations=['United Kingdom'],
                    locationmode="country names",
                    z=[0],
                    showscale=False,
                    marker=dict(
                        line=dict(
                            color='black',
                            width=2
                        )
                    ),
                    colorscale=[[0, 'rgba(255,255,0,0.2)'], [1, 'rgba(255,255,0,0.2)']],
                    name='UK Highlight'
                )
            ],
            name=str(year),
            layout=go.Layout(annotations=annotations),
        )
    )

fig.frames = frames

# Get initial UK countries data for the first annotations
initial_uk_countries = filtered_data[filtered_data['Country'].isin(['Scotland', 'Wales', 'Northern Ireland'])]

# Initial annotations for first display
initial_annotations = [
    dict(
        x=0.01,
        y=0.99,
        xref="paper",
        yref="paper",
        text=(
            f"<b>Year: {initial_year}</b><br>"
            f"Total Players: {int(pl_foreign_percentage.loc['Total Players', initial_year])}<br>"
            f"Foreign Players: {int(pl_foreign_percentage.loc['Foreign Players', initial_year])}<br>"
            f"Percentage: {pl_foreign_percentage.loc['Percentage', initial_year]:.2f}%"
        ),
        showarrow=False,
        align="left",
        font=dict(size=14, color="black"),
        bgcolor="rgba(255, 255, 255, 0.8)",
        bordercolor="black",
        borderwidth=1,
    ),
    # Add UK countries box for initial display
    dict(
        x=0.01,
        y=0.82,
        xref="paper",
        yref="paper",
        text=(
            f"<b>UK countries:</b><br>"
            f"Scotland: {initial_uk_countries[initial_uk_countries['Country'] == 'Scotland']['Players'].sum() if not initial_uk_countries[initial_uk_countries['Country'] == 'Scotland'].empty else 0}<br>"
            f"Wales: {initial_uk_countries[initial_uk_countries['Country'] == 'Wales']['Players'].sum() if not initial_uk_countries[initial_uk_countries['Country'] == 'Wales'].empty else 0}<br>"
            f"Northern Ireland: {initial_uk_countries[initial_uk_countries['Country'] == 'Northern Ireland']['Players'].sum() if not initial_uk_countries[initial_uk_countries['Country'] == 'Northern Ireland'].empty else 0}"
        ),
        showarrow=False,
        align="left",
        font=dict(size=14, color="black"),
        bgcolor="rgba(255, 255, 255, 0.8)",
        bordercolor="black",
        borderwidth=1,
    )
]

# Update layout for better visuals
fig.update_layout(
    geo=dict(
        showframe=False,
        showcoastlines=True,
        showcountries=True,
        coastlinecolor="#444",
        countrycolor="#666",
        landcolor="#f0f0f0",  # Light gray land
        oceancolor="#cfe2f3",  # Light blue ocean
        projection_type="miller",  # Changed to Miller projection
        resolution=110,  # Higher resolution
        showlakes=True,
        lakecolor="#cfe2f3",  # Light blue lakes
    ),
    title=dict(
        text="Number of Foreign Players in the Premier League by Year",
        x=0.5,
        y=0.95,  # Adjusted position
        font=dict(size=20, family="Arial", color="#333")
    ),
    width=1200,  # Increased width
    height=700,  # Increased height
    margin=dict(l=10, r=10, t=60, b=80),  # Reduced margins, but kept space for title and slider
    paper_bgcolor="white",
    updatemenus=[
        dict(
            type="buttons",
            showactive=False,
            x=0.1,
            y=0,
            xanchor="right",
            yanchor="top",
            pad=dict(t=0, r=10),
            buttons=[
                dict(label="Play", method="animate", args=[None, {"frame": {"duration": 800, "redraw": True}, "fromcurrent": True}]),
                dict(label="Pause", method="animate", args=[[None], {"frame": {"duration": 0, "redraw": False}, "mode": "immediate"}]),
            ],
        )
    ],
    # Add initial annotations
    annotations=initial_annotations
)

# Add slider with improved styling
fig.update_layout(
    sliders=[
        dict(
            active=0,
            steps=[
                dict(
                    method="animate",
                    args=[[str(year)], {"frame": {"duration": 800, "redraw": True}, "mode": "immediate"}],
                    label=str(year),
                )
                for year in years
            ],
            transition={"duration": 300},
            x=0.1,
            xanchor="left",
            y=0.01,  # Positioned at bottom with small margin
            yanchor="top",
            pad=dict(b=10, t=10),
            len=0.8,  # Make slider wider
            currentvalue=dict(
                visible=True,
                prefix="Year: ",
                xanchor="right",
                font=dict(size=14, color="#666")
            ),
        )
    ]
)

# Show the map
fig.show()
```
