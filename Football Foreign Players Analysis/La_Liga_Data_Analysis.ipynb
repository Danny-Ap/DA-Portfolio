```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go
```

```python
# Get top countries list of most players in La Liga
la_liga_num = pd.read_csv("La_Liga_Num_Of_Players.csv")
la_liga_num.set_index('Unnamed: 0', inplace=True)
la_liga_num.rename_axis("Country", inplace=True)
la_liga_num_top = la_liga_num.sum(axis=1).sort_values(ascending=False)
la_liga_num_top_list = la_liga_num_top.index.tolist()


# Get top countries list of most minutes played by country in La Liga
la_liga_min = pd.read_csv("La_Liga_Min_Played.csv")
la_liga_min.set_index('Unnamed: 0', inplace=True)
la_liga_min.rename_axis("Country", inplace=True)
la_liga_min_top = la_liga_min.sum(axis=1).sort_values(ascending=False)
la_liga_min_top_list = la_liga_min_top.index.tolist()
```

```python
# Get total number of players per year percentage
la_liga_num_total = la_liga_num.sum(axis=0)
la_liga_num_percentage = la_liga_num.copy()
for year in la_liga_num.columns:
    la_liga_num_percentage[year] = (la_liga_num[year] / la_liga_num_total[year] * 100).round(2)

la_liga_num_percentage
```

```python
# Get total minutes played per year percentage
la_liga_min_total = la_liga_min.sum(axis=0)
la_liga_min_percentage = la_liga_min.copy()
for year in la_liga_min.columns:
    la_liga_min_percentage[year] = (la_liga_min[year] / la_liga_min_total[year] * 100).round(2)

la_liga_min_percentage
```

```python
# Select the top 10 countries (skip Spain, 1st row)
la_liga_num_top_10 = la_liga_num.loc[la_liga_num_top_list[1:11]]

# Plot each country's data
plt.figure(figsize=(12, 8))
for index, row in la_liga_num_top_10.iterrows():
    plt.plot(row.index, row.values, label=index)
    plt.annotate(index, xy=(row.index[-1], row.values[-1]), xytext=(5, 0), textcoords='offset points')

# Add labels and title
plt.xlabel('Year')
plt.xticks(rotation=90)
plt.ylabel('Number of Players')
plt.title('Number of Players by Country Over Years')
plt.grid(axis='both',linestyle='--', alpha=0.7)
plt.yticks(range(0, int(la_liga_num_top_10.max().max())+1, 5))

# Show the plot
plt.show()
```

```python
# Select the top 10 countries (skip Italy, 1st row)
la_liga_num_percentage_top_10 = la_liga_num_percentage.loc[la_liga_num_top_list[1:11]]

# Plot each country's data
plt.figure(figsize=(12, 8))
for index, row in la_liga_num_percentage_top_10.iterrows():
    plt.plot(row.index, row.values, label=index)
    plt.annotate(index, xy=(row.index[-1], row.values[-1]), xytext=(5, 0), textcoords='offset points')

# Add labels and title
plt.xlabel('Year')
plt.xticks(rotation=90)
plt.ylabel('Percentage')
plt.title('Percentage of Foreign Players by Country Over Years')
plt.grid(axis='both',linestyle='--', alpha=0.7)
plt.yticks(range(0, int(la_liga_num_percentage_top_10.max().max())+1, 1))

# Show the plot
plt.show()
```

```python
df1 = pd.read_csv('La_Liga_Num_Of_Players.csv')
df1.set_index('Unnamed: 0', inplace=True)
df1.loc['Total Players'] = df1.sum()
la_liga_total_num_of_players = df1.loc[['Total Players']]

df2 = pd.read_csv('La_Liga_Num_Of_Players.csv')
df2.set_index('Unnamed: 0', inplace=True)
df2.loc['Foreign Players'] = df2.iloc[1:].sum()
la_liga_foreign_players_number = df2.loc[['Foreign Players']]

la_liga_foreign_percentage = la_liga_total_num_of_players.copy()
la_liga_foreign_percentage.loc['Foreign Players'] = la_liga_foreign_players_number.iloc[0]
percentage_row = (la_liga_foreign_percentage.loc['Foreign Players'] / la_liga_foreign_percentage.loc['Total Players'] * 100).round(2)
la_liga_foreign_percentage.loc['Percentage'] = percentage_row
la_liga_foreign_percentage
```

```python
# Cleaning df for worldmap scatter plot

la_liga_foreign_countries = pd.read_csv("La_Liga_Num_Of_Players.csv")
la_liga_foreign_countries.set_index('Unnamed: 0', inplace=True)
la_liga_foreign_countries.rename_axis("Country", inplace=True)

la_liga_foreign_countries = la_liga_foreign_countries.iloc[1:]  # Remove Italy row
la_liga_foreign_countries.index = la_liga_foreign_countries.index.str.extract(r'([A-Z][a-zA-Z\s]*)$')[0]  # Extract country names

la_liga_foreign_countries_long = la_liga_foreign_countries.reset_index(names='Country').melt(
    id_vars='Country', var_name='Year', value_name='Players'
)

la_liga_foreign_countries_long
```

```python
# Create the base figure
fig = go.Figure()

# Find global maximum for consistent color scale
global_max = la_liga_foreign_countries_long['Players'].max()

# Add the initial choropleth map for the first year
initial_year = la_liga_foreign_countries_long['Year'].min()
filtered_data = la_liga_foreign_countries_long[la_liga_foreign_countries_long['Year'] == initial_year]

fig.add_trace(
    go.Choropleth(
        locations=filtered_data['Country'],
        locationmode="country names",
        z=filtered_data['Players'],
        text=filtered_data['Country'],
        colorscale="Reds",  # Changed from ["white","red"] to "Reds"
        colorbar_title="Players",
        zmin=0,
        zmax=global_max,  # Set global maximum
        showscale=True,    # Show the color scale
    )
)

# Add frames for each year
years = la_liga_foreign_countries_long['Year'].unique()
frames = []
for year in years:
    year_data = la_liga_foreign_countries_long[la_liga_foreign_countries_long['Year'] == year]
    annotations = [
        dict(
            x=0.01,  # Moved to left side
            y=0.99,  # Moved to top
            xref="paper",
            yref="paper",
            text=(
                f"<b>Year: {year}</b><br>"
                f"Total Players: {int(la_liga_foreign_percentage.loc['Total Players', year])}<br>"
                f"Foreign Players: {int(la_liga_foreign_percentage.loc['Foreign Players', year])}<br>"
                f"Percentage: {la_liga_foreign_percentage.loc['Percentage', year]:.2f}%"
            ),
            showarrow=False,
            align="left",
            font=dict(size=14, color="black"),
            bgcolor="rgba(255, 255, 255, 0.8)",  # Semi-transparent background
            bordercolor="black",
            borderwidth=1,
        )
    ]
    frames.append(
        go.Frame(
            data=[
                go.Choropleth(
                    locations=year_data['Country'],
                    locationmode="country names",
                    z=year_data['Players'],
                    text=year_data['Country'],
                    colorscale="Reds",  # Changed from ["white","red"] to "Reds"
                    zmin=0,
                    zmax=global_max,  # Set global maximum
                    showscale=True,
                )
            ],
            name=str(year),
            layout=go.Layout(annotations=annotations),
        )
    )

fig.frames = frames

# Update layout for better visuals
fig.update_layout(
    geo=dict(
        showframe=False,
        showcoastlines=True,
        showcountries=True,
        coastlinecolor="#444",
        countrycolor="#666",
        landcolor="#f0f0f0",  # Light gray land
        oceancolor="#cfe2f3",  # Light blue ocean
        projection_type="miller",  # Changed to Miller projection
        resolution=110,  # Higher resolution
        showlakes=True,
        lakecolor="#cfe2f3",  # Light blue lakes
    ),
    title=dict(
        text="Number of Foreign Players in La Liga by Year",
        x=0.5,
        y=0.95,  # Adjusted position
        font=dict(size=20, family="Arial", color="#333")
    ),
    width=1200,  # Increased width
    height=700,  # Increased height
    margin=dict(l=10, r=10, t=60, b=80),  # Reduced margins, but kept space for title and slider
    paper_bgcolor="white",
    updatemenus=[
        dict(
            type="buttons",
            showactive=False,
            x=0.1,
            y=0,
            xanchor="right",
            yanchor="top",
            pad=dict(t=0, r=10),
            buttons=[
                dict(label="Play", method="animate", args=[None, {"frame": {"duration": 800, "redraw": True}, "fromcurrent": True}]),
                dict(label="Pause", method="animate", args=[[None], {"frame": {"duration": 0, "redraw": False}, "mode": "immediate"}]),
            ],
        )
    ],
)

# Add slider with improved styling
fig.update_layout(
    sliders=[
        dict(
            active=0,
            steps=[
                dict(
                    method="animate",
                    args=[[str(year)], {"frame": {"duration": 800, "redraw": True}, "mode": "immediate"}],
                    label=str(year),
                )
                for year in years
            ],
            transition={"duration": 300},
            x=0.1,
            xanchor="left",
            y=0.01,  # Positioned at bottom with small margin
            yanchor="top",
            pad=dict(b=10, t=10),
            len=0.8,  # Make slider wider
            currentvalue=dict(
                visible=True,
                prefix="Year: ",
                xanchor="right",
                font=dict(size=14, color="#666")
            ),
        )
    ]
)

# Show the map
fig.show()
```
