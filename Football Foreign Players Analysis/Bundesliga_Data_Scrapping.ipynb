```python
# Data scrapping
import pandas as pd
```

```python
# Year parameter changeable to get csv for every year from 1988-1989 to 2023-2024
year = 2023
url = f"https://fbref.com/en/comps/20/{year}-{year+1}/nations/{year}-{year+1}-Bundesliga-Nationalities"
df = pd.read_html(url, attrs={"id":"nations"})[0]
df.to_csv(f"nat{year}.csv")
```

```python
# For current season (2024-2025), the url is different
url = "https://fbref.com/en/comps/20/nations/Bundesliga-Nationalities"
df = pd.read_html(url, attrs={"id":"nations"})[0]
df.to_csv("nat2024.csv")
```

```python
# DATA CLEANING
year = 2024
df = pd.read_csv(f"nat{year}.csv")
df.set_index("Rk", inplace=True)
df.drop(columns="Unnamed: 0", inplace=True)
df.drop('Rk', inplace=True)
df.to_csv(f"nat{year}.csv")
```

```python
# Makes a new df, with rows as countries, and columns as years. The values are the number of players from that country during that year

Bundesliga_Num_Of_Players = pd.DataFrame()
years = range(1988,2025)

for year in years:
    df = pd.read_csv(f'Bundesliga_Nationalities\\nat{year}.csv')

    year_col = str(year)
    Bundesliga_Num_Of_Players[year_col] = 0

    for index, row in df.iterrows():
        country = row['Nation']
        players = row['# Players']

        if country not in Bundesliga_Num_Of_Players.index: # if during that year no player from that country played in the Serie A, then the value is 0
            Bundesliga_Num_Of_Players.loc[country] = 0
        Bundesliga_Num_Of_Players.at[country, year_col] = players

Bundesliga_Num_Of_Players.to_csv("Bundesliga_Num_Of_Players.csv")
```

```python
# Makes a new df, with rows as countries, and columns as years. The values are the minutes played by players from that country during that year

Bundesliga_Min_Played = pd.DataFrame()
years = range(1988,2025)

for year in years:
    df = pd.read_csv(f'Bundesliga_Nationalities\\nat{year}.csv')

    year_col = str(year)
    Bundesliga_Min_Played[year_col] = 0

    for index, row in df.iterrows():
        country = row['Nation']
        minutes = row['Min']

        if country not in Bundesliga_Min_Played.index: # if during that year no player from that country played in the Serie A, then the value is 0
            Bundesliga_Min_Played.loc[country] = 0
        Bundesliga_Min_Played.at[country, year_col] = minutes

Bundesliga_Min_Played.to_csv("Bundesliga_Min_Played.csv")
```
